```python
import random

print("=======================================")
print("SELAMAT DATANG DI DUNGEON MATH1011")
print("=======================================\n")
print("Kalahkan monster angka untuk lulus mata kuliah ini!")

skor = 0
total_soal_biasa = 10

# ==========================================
# STAGE 1: 10 PERTANYAAN BIASA (MONSTER KECIL)
# ==========================================
for i in range(1, total_soal_biasa + 1):
    # 1. Random Nama Depan (tanpa list)
    rand_nama = random.randint(1, 6)
    if rand_nama == 1:
        nama_depan = "Wilbert"
    elif rand_nama == 2:
        nama_depan = "Christoph"
    elif rand_nama == 3:
        nama_depan = "Lionel"
    elif rand_nama == 4:
        nama_depan = "Joel"
    elif rand_nama == 5:
        nama_depan = "Evan"
    else:
        nama_depan = "Yosia"

    # 2. Random Istilah Kalkulus (tanpa list)
    rand_materi = random.randint(1, 6)
    if rand_materi == 1:
        materi = "Garis Singgung"
    elif rand_materi == 2:
        materi = "Integral"
    elif rand_materi == 3:
        materi = "Turunan"
    elif rand_materi == 4:
        materi = "Limit"
    elif rand_materi == 5:
        materi = "Asimtot"
    else:
        materi = "Diferensial"
        
    # Gabungkan nama monster
    monster = nama_depan + " " + materi
        
    print(f"\n--- [Lantai {i}] Monster {monster} menghadangmu! ---")
    
    # 3. Random operator (1: +, 2: -, 3: *, 4: /)
    op_random = random.randint(1, 4)
    
    if op_random == 1:
        operator = '+'
        angka1 = random.randint(1, 20)
        angka2 = random.randint(1, 20)
        jawaban_asli = angka1 + angka2
    elif op_random == 2:
        operator = '-'
        angka1 = random.randint(10, 30)
        angka2 = random.randint(1, 10) 
        jawaban_asli = angka1 - angka2
    elif op_random == 3:
        operator = '*'
        angka1 = random.randint(1, 10)
        angka2 = random.randint(1, 10)
        jawaban_asli = angka1 * angka2
    else:
        operator = '/'
        angka2 = random.randint(2, 10)
        jawaban_asli = random.randint(2, 12)
        angka1 = angka2 * jawaban_asli

    # 4. Random 3 pengecoh biar total jadi 4 pilihan
    pengecoh1 = jawaban_asli + random.randint(1, 3)
    pengecoh2 = jawaban_asli - random.randint(1, 3)
    pengecoh3 = jawaban_asli + random.randint(4, 7)
    
    # 5. Random posisi pilihan ganda (1 sampai 4)
    posisi_benar = random.randint(1, 4)
    
    print(f"Serang titik lemahnya dengan menjawab: {angka1} {operator} {angka2} = ?")
    
    # Menampilkan 4 pilihan ganda murni pakai if-else tanpa list
    if posisi_benar == 1:
        print(f"a. {jawaban_asli}\nb. {pengecoh1}\nc. {pengecoh2}\nd. {pengecoh3}")
        kunci = 'a'
    elif posisi_benar == 2:
        print(f"a. {pengecoh1}\nb. {jawaban_asli}\nc. {pengecoh2}\nd. {pengecoh3}")
        kunci = 'b'
    elif posisi_benar == 3:
        print(f"a. {pengecoh1}\nb. {pengecoh2}\nc. {jawaban_asli}\nd. {pengecoh3}")
        kunci = 'c'
    else:
        print(f"a. {pengecoh1}\nb. {pengecoh2}\nc. {pengecoh3}\nd. {jawaban_asli}")
        kunci = 'd'
        
    # 6. Input & Pengecekan
    jawaban_user = input("Pilihanmu (a/b/c/d): ").lower()
    
    if jawaban_user == kunci:
        print(f"CRITICAL HIT! {monster} hancur. (+10 Poin)")
        skor += 10
    else:
        print(f"Serangan meleset! Yang benar adalah {kunci}.")

# ==========================================
# STAGE 2: FINAL BOSS (Sanga)
# ==========================================
print("\n=======================================")
print("WARNING: FINAL BOSS APPROACHING!")
print("=======================================")
print("Sanga, sang Dosen MATH1011 turun ke medan perang!")

# Logika Final Boss
total_putaran = random.randint(40, 60)
while total_putaran % 4 == 0: 
    total_putaran = random.randint(40, 60)

dmg_player = random.randint(1500, 2500)
heal_boss = random.randint(200, 500)

hp_calc = 0
for r in range(1, total_putaran + 1):
    if r % 4 == 0:
        hp_calc += heal_boss 
    else:
        hp_calc += (dmg_player - heal_boss) 

hp_boss = hp_calc - random.randint(1, (dmg_player - heal_boss) - 1)
jawaban_boss_asli = total_putaran

print(f"HP Boss: {hp_boss:,}")
print(f"DMG Senjatamu: {dmg_player:,} / putaran.")
print(f"Regen Boss: {heal_boss:,} HP tiap diserang.")
print(f"Skill Pasif Boss (Divine Shield): Tiap kelipatan 4 putaran, DMG-mu jadi 0, tapi Boss tetap Regen!")
print("\nBerapa putaran yang kamu butuhin buat habisin HP Boss sampai 0?")

# 3 Pengecoh untuk Final Boss (total 4 opsi)
pb1 = jawaban_boss_asli + random.randint(1, 2)
pb2 = jawaban_boss_asli - random.randint(1, 2)
pb3 = jawaban_boss_asli + random.randint(3, 5)

posisi_boss = random.randint(1, 4)

if posisi_boss == 1:
    print(f"a. {jawaban_boss_asli}\nb. {pb1}\nc. {pb2}\nd. {pb3}")
    kunci_boss = 'a'
elif posisi_boss == 2:
    print(f"a. {pb1}\nb. {jawaban_boss_asli}\nc. {pb2}\nd. {pb3}")
    kunci_boss = 'b'
elif posisi_boss == 3:
    print(f"a. {pb1}\nb. {pb2}\nc. {jawaban_boss_asli}\nd. {pb3}")
    kunci_boss = 'c'
else:
    print(f"a. {pb1}\nb. {pb2}\nc. {pb3}\nd. {jawaban_boss_asli}")
    kunci_boss = 'd'

jawaban_user_boss = input("Pilihanmu (a/b/c/d): ").lower()

if jawaban_user_boss == kunci_boss:
    print("\nFATALITY! Sanga telah dikalahkan! (+911 Poin)")
    skor += 911
else:
    print(f"\nYOU DIED. Jawaban yang benar harusnya {kunci_boss} putaran.")

# ==========================================
# STAGE 3: PERHITUNGAN KELULUSAN
# ==========================================
print("\n=======================================")
print("REKAPITULASI NILAI MATH1011")
print(f"Skor Akhir Kamu: {skor}")

if skor == 1011:
    print("Predikat: LULUS MATKUL MATH1011")
    print("Jancokkk, mantap bro! Lu bener-bener Legend.")
elif 70 <= skor <= 100:
    print("Predikat: Lulus Kuis Dasar")
    print("Not bad. Lu lulus kuis dasar, tapi masih belum cukup kuat lawan Boss.")
elif 100 < skor < 1011:
    print("Predikat: Lulus Kuis Dasar (Anomali)")
    print("Lu berhasil bunuh Boss, tapi anehnya monster kroco ada yang lu skip. Tetep diitung lulus dasar aja bro.")
else:
    print("Predikat: TIDAK LULUS")
    print("Cupu, ngulang tahun depan bro.")
```
